```python
import os
import shutil

def move_images_to_final(source_dir, final_dir):
    """
    Move all images from subdirectories of the source directory to a single folder named 'Final'.

    :param source_dir: The root directory containing subdirectories with images.
    :param final_dir: The target directory where all images will be moved.
    """
    # Ensure the Final directory exists
    if not os.path.exists(final_dir):
        os.makedirs(final_dir)

    # Supported image extensions
    image_extensions = {'.png', '.jpg', '.jpeg', '.gif', '.bmp', '.tiff'}

    # Traverse the source directory
    for root, dirs, files in os.walk(source_dir):
        for file in files:
            # Check if the file is an image
            if os.path.splitext(file)[1].lower() in image_extensions:
                source_path = os.path.join(root, file)
                destination_path = os.path.join(final_dir, file)

                # Avoid overwriting files with the same name
                if os.path.exists(destination_path):
                    base, ext = os.path.splitext(file)
                    counter = 1
                    while os.path.exists(destination_path):
                        destination_path = os.path.join(final_dir, f"{base}_{counter}{ext}")
                        counter += 1

                # Move the file
                shutil.move(source_path, destination_path)
                print(f"Moved: {source_path} -> {destination_path}")

# Usage example
source_directory = "./data/small_scale/dead_leaves-squares/train"
final_directory = "./data/small_scale/dead_leaves-squares/Final"

move_images_to_final(source_directory, final_directory)
```
