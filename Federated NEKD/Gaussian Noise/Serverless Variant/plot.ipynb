```python
import pandas
from matplotlib import pyplot as plt
```

```python
def plot_figures(file_path,title):
    df = pandas.read_csv(file_path)
    df = df.T
    df.columns = df.iloc[0]
    df = df.drop(df.index[0])
    df = df.astype(float)
    epochs = list(range(1, len(df)+1))
    fig, ax = plt.subplots(1,3,figsize=(20,5))
    fig.suptitle(title)
    ax[0].plot(epochs,df['Accuracy'],label='Accuracy')
    ax[0].set_xlabel('Epoch')
    ax[0].set_ylabel('Accuracy')
    ax[0].set_title('Accuracy vs Epoch')
    ax[0].legend()
    ax[0].grid(True)
    ax[1].plot(epochs,df['F1 Score'],label='F1 Score')
    ax[1].set_xlabel('Epoch')
    ax[1].set_ylabel('F1 Score')
    ax[1].set_title('F1 Score vs Epoch')
    ax[1].legend()
    ax[1].grid(True)
    ax[2].plot(epochs,df['Variance Grad'],label='Variance Grad')
    ax[2].set_xlabel('Epoch')
    ax[2].set_ylabel('Variance of Gradient')
    ax[2].set_title('Variance of Gradient vs Epoch')
    ax[2].legend()
    ax[2].grid(True)
    plt.show()
```

```python
plot_figures('Results/CIFAR10_10_42_Drichlet_0.100_default__NEKD_SGD_sgd.csv','NEKD_SGD Serverless ($\lambda_1 = 1, \lambda_2=0$) - Only Client Data')
```
